```python
import pandas as pd
from sklearn.datasets import load_iris

# Load the iris dataset
iris = load_iris()

# Create a DataFrame using the data and feature names from the iris dataset
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df_iris['target'] = iris.target

# Display the DataFrame
print(df_iris.head())
```

```python
df_iris.columns
```

```python
df_iris.rename({'target':'species'}, axis=1, inplace=True) ##Renombrar la columna target
```

```python
df_iris
```

```python
df_iris['species'] = df_iris['species'].astype('category')
df_iris.info()
```

```python
iris.amp = mf.ampute_data(df_iris, perc=0.25, random_state=100) ##asigno valores nulos a las columnas
iris.amp
```

```python
##aplicando el imputacionkernel
kds = mf.ImputationKernel(
    iris.amp,
    random_state=100  ##valores aleatorios para entrenar modelo
)
```

```python
kds.mice(2) ## correr el algoritmo MICE durante 2 iteraciones
```

```python
iris_completo=kds.complete_data()
iris_completo
```

```python
#Autocorrelación de la serie de tiempo - Que el día anterior afecta al día siguiente
from statsmodels.graphics.tsaplots import plot_acf
df_fecha_agr = df_fecha.groupby('fecha').size()

plt.figure(figsize=(12,6))
plot_acf(df_fecha_agr, lags=365)
plt.xlabel('Lags')
plt.ylabel('Autocorrelación')
plt.title('Autocorrelation Function (ACF) Plot')
plt.show()
```
